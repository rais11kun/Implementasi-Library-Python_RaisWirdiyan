# Teorema Bayes (Sistem Deteksi Spam)**teks tebal**

```python
P_S = 0.3
P_not_S = 0.7
P_D_given_S = 0.8
P_D_given_not_S = 0.1

# Hukum Probabilitas Total (Evidence)
P_D = (P_D_given_S * P_S) + (P_D_given_not_S * P_not_S)

# Teorema Bayes (Posterior Probability)
P_S_given_D = (P_D_given_S * P_S) / P_D

print(f"Probabilitas Total Kata 'Diskon' muncul P(D) = {P_D:.3f}")
print(f"Probabilitas email adalah Spam jika ada kata 'Diskon' P(S|D) = {P_S_given_D:.3f}")
```

# Simulasi Kejadian Acak (Uptime Server)**teks tebal**

```python
import numpy as np
import matplotlib.pyplot as plt

server = ['Server A', 'Server B', 'Server C']
probabilitas = [0.5, 0.3, 0.2]
jumlah_request = 10000

# Menggunakan np.random.choice untuk probabilitas berbobot skala besar
hasil_simulasi = np.random.choice(server, size=jumlah_request, p=probabilitas)

count_A = np.sum(hasil_simulasi == 'Server A')
count_B = np.sum(hasil_simulasi == 'Server B')
count_C = np.sum(hasil_simulasi == 'Server C')

print(f"Server A: {count_A} request")
print(f"Server B: {count_B} request")
print(f"Server C: {count_C} request")

plt.bar(server, [count_A, count_B, count_C], color=['blue', 'orange', 'green'])
plt.title('Simulasi Load Balancing 10.000 Request')
plt.ylabel('Jumlah Request')
plt.show()
```

# **Distribusi Diskrit - Binomial (Pengiriman Paket Jaringan)**

```python
from scipy.stats import binom
import matplotlib.pyplot as plt
import numpy as np

n, p = 20, 0.05
# PMF mencari fungsi probabilitas diskrit kejadian mutlak P(X = 2)
prob_2_gagal = binom.pmf(2, n, p)
print(f"Probabilitas tepat 2 paket gagal: {prob_2_gagal:.4f}")

x = np.arange(0, 11)
y = binom.pmf(x, n, p)
plt.vlines(x, 0, y, colors='b', lw=5, alpha=0.5)
plt.plot(x, y, 'bo', ms=8)
plt.title('Distribusi Binomial Paket Drop (n=20, p=0.05)')
plt.xlabel('Jumlah Paket yang Gagal/Drop')
plt.ylabel('Probabilitas Kejadian (PMF)')
plt.show()
```

# **Distribusi Kontinu - Normal (Waktu Respons API)**

```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt

mu, sigma = 120, 15
# CDF menjumlahkan fungsi densitas peluang luasan area dari kiri hingga batas P(X <= 100)
prob_under_100 = norm.cdf(100, mu, sigma)
print(f"Probabilitas response time API di bawah 100 ms: {prob_under_100:.4f}")

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)
plt.plot(x, y, label='Densitas Kurva Lonceng')

# Visualisasi blok integral area kumulatif CDF
plt.fill_between(x, y, where=(x < 100), color='red', alpha=0.3, label='Area < 100ms')
plt.title('Distribusi Normal Response Time API')
plt.xlabel('Waktu Respons (ms)')
plt.ylabel('Kepadatan Probabilitas Densitas (PDF)')
plt.legend()
plt.show()
```
